## **3. Seeing models as computation graphs**

Following the tensor library, the second pillar of PyTorch is *Autograd*, an automatic differentiation engine that operates using computational graphs.

The Computational Graph is a directed graph that visualizes and organizes mathematical expressions as a sequence of operations. It maps the "forward pass" (prediction step) of a model, tracking how inputs and parameters flow through functions to produce an output or loss. PyTorch builds this graph in the background to enable backpropagation, the process of calculating gradients required to optimize model parameters.

### **Example: Logistic Regression**

In a simple logistic regression classifier, the graph connects input features ($x$), weights ($w$), and biases ($b$) to compute an activation ($a$) and a final loss. This structured sequence allows PyTorch to automatically determine how changing a weight affects the total error.

<figure style='text-align: center'>
<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/7.webp" width="800px">
</figure>

```python
import torch.nn.functional as F

y = torch.tensor([1.0])  # true label
x1 = torch.tensor([1.1]) # input feature
w1 = torch.tensor([2.2]) # weight parameter
b = torch.tensor([0.0])  # bias unit

z = x1 * w1 + b          # net input
a = torch.sigmoid(z)     # activation & output

loss = F.binary_cross_entropy(a, y)
print(loss)
```

## **4 Automatic differentiation made easy**

PyTorch’s Autograd engine automates the complex calculus required to train neural networks, specifically by calculating gradients through a process called backpropagation.

<figure style='text-align: center'>
<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/8.webp" width="800px">
</figure>

By setting `requires_grad=True` on a tensor (like weights or biases), you tell PyTorch to track all operations involving that tensor to build a computation graph. PyTorch uses *reverse-mode automatic differentiation* to apply the chain rule from the loss back to the input, determining how each parameter should be updated to minimize error. To run backpropagation manually, we run `grad(objective, param)`. By default, PyTorch destroys the computation graph after calculating the gradients to free memory. However, since we are going to reuse this computation graph with `.backward()` method, we set `retain_graph=True` inside `grad()` function so that it stays in memory.

Instead of manual derivation, calling `.backward()` on a loss tensor automatically calculates gradients for all relevant parameters and stores them in their respective `.grad` attributes for all leaves of the computational graph. While you can use the `grad()` function for manual experimentation, the standard workflow in PyTorch leverages the `.backward()` method to handle the math automatically.

> **Takeaway:** You don't need to perform manual calculus; PyTorch tracks the operations and computes the derivatives for you with a single method call.

```python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

z = x1 * w1 + b 
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a, y)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print(grad_L_w1)
print(grad_L_b)
```

```python
loss.backward()

print(w1.grad)
print(b.grad)
```

## **5. Implementing multilayer neural networks**

The third pillar of PyTorch is its Deep Learning Library, centered around the `torch.nn.Module` class, which provides the framework for building complex architectures like Multilayer Perceptrons (MLPs).

<figure style='text-align: center'>
<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/9.webp" width="600px">
</figure>

* Subclassing `torch.nn.Module` allows you to bundle layers, operations, and parameters into a single, manageable object. The `torch.nn.Module` class (the parent) contains complex internal logic for tracking gradients, registering parameters, and handling GPU/CPU moves. If you don't call `super().__init__()`, your child class is just an empty shell.
* The Constructor (`__init__`) is used to define the structural building blocks of the network, such as fully connected (linear) layers or activation functions.
* The forward pass (`forward`) defines the data flow from input to output. When you call this method, PyTorch automatically constructs the computation graph required for training.
* When you define the flow in the `forward` method, PyTorch’s autograd engine automatically handles the `backward` pass (gradient calculation) without requiring manual coding.

When implementing a network, you focus on defining **what** the layers are and **how** data moves through them; the library handles the underlying parameter tracking and calculus.

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
                
            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
```

We can then instantiate a new neural network object. It is often useful to call print on the model to see a summary of its structure:

```python
model = NeuralNetwork(50, 3)
print(model)
```

The `torch.nn.Sequential` container is a productivity tool used to group layers that execute in a fixed, linear order. Here are key advantages of sequenntial layer implementation:

* **Code Simplicity:** Instead of calling each layer individually (e.g., `x = self.layer1(x)`, `x = self.relu(x)`, etc.) inside the `forward` method, you simply call the `Sequential` object once.
* **Organization:** It consolidates the architecture within the `__init__` constructor, making the network's structure easier to read and maintain.

While convenient for linear architectures, `Sequential` is less suitable for complex models requiring non-linear data flows, such as skip connections or multiple inputs/outputs. We can rewrite `NeuralNetwork` class using the functional API as below:

```python
import torch.nn.functional as F

class NeuralNetworkFunctional(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        # Instantiate functional calls for three layers    
        self.fc1 = torch.nn.Linear(num_inputs, 30)
        self.fc2 = torch.nn.Linear(30, 20)
        self.fc3 = torch.nn.Linear(20, num_outputs)
            
    def forward(self, x):
        
        # 1st layer
        x = self.fc1(x)
        x = F.relu(x)
        
        # 2nd layer
        x = self.fc2(x)
        x = F.relu(x)
        
        # Output logits
        logits = self.fc3(x)         
        return logits
```

```python
# Create dummy input data (Batch Size = 5, Features = 50)
x = torch.rand((10, 50))
print(x)
```

In PyTorch, every time you instantiate a `torch.nn.Linear` layer, the weights and biases are initialized with small random numbers. Even if the code looks the same, the "starting point" for those numbers is different every time. To compare the outputs of the same models, we must set a "seed" for the random number generator before creating each model:

```python
# Set seed for sequential model 
torch.manual_seed(123)
model_sequential = NeuralNetwork(50, 3)

# Set seed for functional model
torch.manual_seed(123)
model_functional = NeuralNetworkFunctional(50, 3)

assert torch.allclose(model_sequential(x), model_functional(x), atol=1e-6)
```

As briefly mentioned above, deep learning models use random weight initialization to "break symmetry," ensuring that different neurons learn unique features rather than performing identical calculations during training. If all weights started at the same value (e.g., zero), every neuron would produce the same output and receive the same gradient update, preventing the model from learning complex patterns. While randomness is essential for learning, it can make debugging difficult. You can make results predictable by using `torch.manual_seed()`, which forces the random number generator to produce the same "random" sequence every time.

```python
torch.manual_seed(123)

out = model(x)
print(out)
```

When you call `model(x)`, PyTorch executes the forward pass, transforming input features into output scores (logits) through each layer of the network. The computational graph tracks inputs flow from the input layer through hidden layers to the output. In this process, each resulting tensor contains a `grad_fn` attribute (e.g., `<AddmmBackward0>`). This "footprint" identifies the mathematical operations used, like matrix multiplication (`mm`) and addition (`Add`), so PyTorch can retrace its steps during backpropagation.

During inference, however, tracking gradients is unnecessary and resource-intensive. Wrapping your code in the `torch.no_grad()` context manager disables the computational graph. This reduces: (1) memory consumption by not storing intermediate tensors, and (2) speeds up computation by skipping gradient-related overhead.

```python
with torch.no_grad():
    out = model(x)
print(out)
```

In PyTorch, it is standard practice for models to return **raw logits** (the direct outputs of the final linear layer) rather than probabilities. The main reason is numerical stability. Common PyTorch loss functions (like `CrossEntropyLoss`) internally combine a nonlinear activation (Softmax/Sigmoid) with the loss calculation. This fused approach is more mathematically stable and efficient. Another reason is flexibility during training and inference. During training, you feed raw logits directly into the loss function while during inference you must explicitly apply `torch.softmax()` to the output to get human-readable class-membership probabilities (which sum to 1).

```python
with torch.no_grad():
    out = torch.softmax(model(x), dim=1)
print(out)
```

## **6. Setting up efficient data loaders**

Efficient data loading in PyTorch relies on two specialized classes that separate the storage of data (`Dataset` class) from the logic of accessing it (`DataLoader` class): 

* **The `Dataset` Class:** Acts as the container for your raw data. It defines how to load an individual record and its corresponding label (e.g., loading an image from a file path).
* **The `DataLoader` Class:** Acts as the engine. It handles how data is served to the model by managing:
    * **Batching:** Grouping individual records into chunks (batches).
    * **Shuffling:** Randomizing the order of data to prevent the model from learning patterns based on sequence.
    * **Parallelization:** Using multiple CPU cores to load data while the GPU is training.

<figure style='text-align: center'>
<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/10.webp" width="800px">
</figure>

To prepare for training, you typically create the dataset (features and labels) and split it into two distinct sets:

1. **Training Set:** Used by the model to learn weights.
2. **Test Set:** Used to evaluate the model's performance on unseen data.

Let’s start by creating a simple toy dataset of five training examples with two features each. Accompanying the training examples, we also create a tensor containing the corresponding class labels: three examples belong to class 0, and two examples belong to class 1. In addition, we also make a test set consisting of two entries. The code to create this dataset is shown below.

```python
X_train = torch.tensor([
    [-1.2, 3.1],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5]
])

y_train = torch.tensor([0, 0, 0, 1, 1])
```

```python
X_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])

y_test = torch.tensor([0, 1])
```

```python
from torch.utils.data import Dataset


class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]        
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = ToyDataset(X_train, y_train)
test_ds = ToyDataset(X_test, y_test)
```

```python
print(train_ds[1])
print(len(train_ds))
```

A custom PyTorch Dataset serves as a bridge between raw data and the training process, structured around three required methods:

* **`__init__` (Setup):** Initializes the data sources. While here it simply stores tensors in memory, it can also be used to define file paths or database connectors.
* **`__getitem__` (Fetch):** Defines how to retrieve a single record (feature and label) using an `index`. This is the method the `DataLoader` calls repeatedly to assemble a batch.
* **`__len__` (Count):** Returns the total number of items in the dataset (e.g., the number of rows in your label tensor). This tells the training loop when a full pass through the data (epoch) is complete.

The next step is passing these `Dataset` objects into a `DataLoader` to automate the batching and shuffling process.

```python
from torch.utils.data import DataLoader

torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0
)
```

```python
test_ds = ToyDataset(X_test, y_test)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False,
    num_workers=0
)
```

```python
for idx, (x, y) in enumerate(train_loader):
    print(f"Batch {idx+1}:", x, y)
```

Iterating through the `DataLoader` defines a *training epoch*, where every example in the dataset is seen exactly once. Here are some of the main features of DataLoader:

* **Dynamic Shuffling:** While a manual seed ensures initial reproducibility, the `DataLoader` reshuffles data differently in each subsequent epoch. This prevents repetitive update cycles, helping the model generalize better.
* **Handling Incomplete Batches:** If the dataset size isn't perfectly divisible by the `batch_size`, the final batch will be smaller (e.g., a batch of 1 when the size is 2). To prevent these small, "noisy" batches from destabilizing training convergence, it is best practice to drop the final incomplete batch using the `drop_last=True` parameter.

```python
train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0,
    drop_last=True,
)
```

```python
for idx, (x, y) in enumerate(train_loader):
    print(f"Batch {idx+1}:", x, y)
```

The `num_workers` parameter in the `DataLoader` controls how many CPU processes are dedicated to loading and preprocessing data in the background. Wehn we set `num_workers=0`, data loading happens in the main process. This creates a bottleneck where the GPU sits idle because it must wait for the CPU to finish fetching the next batch before it can resume training. However, when `num_workers > 0`, multiple worker processes load data in parallel, pre-fetching batches so they are ready the moment the GPU finishes its current task.

For tiny datasets or toy examples, setting `num_workers > 0` can actually slow things down due to the overhead of starting new processes. In interactive environments like Jupyter Notebooks, high worker counts can occasionally cause resource-sharing errors or crashes. While the ideal number depends on your CPU cores and hardware, `num_workers=4` is a common starting point for optimal performance on real-world datasets.

## **7. A typical training loop**



```python
list(train_loader)
```

```python
import torch.nn.functional as F

torch.manual_seed(123)

model = NeuralNetwork(num_inputs=2, num_outputs=2)
optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

num_epochs = 3
for epoch in range(num_epochs):

    model.train()
    for batch_idx, (features, labels) in enumerate(train_loader):
        
        logits = model(features)
        loss = F.cross_entropy(logits, target=labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        print(f"Epoch: {epoch+1:03d}/{num_epochs:03d} | "
              f"Batch: {batch_idx+1:03d}/{len(train_loader)} | "
              f"loss: {loss:.3f}"
        )
```

The code snippet above demonstrates a complete neural network optimization step, relying on specific configurations and practices to ensure stable convergence.

#### **Key Hyperparameters**
* **Learning Rate (`lr`):** A tunable setting that dictates the step size taken toward minimizing the loss.
* **Epochs:** The total number of complete passes through the training dataset.
* **Validation Set:** A separate dataset used repeatedly to fine-tune these hyperparameters without biasing the final evaluation on the test set.

#### **Training vs. Evaluation Modes**

`model.train()` and `model.eval()` toggle specific behaviors for layers like Dropout and Batch Normalization. Though optional for simple networks, including them prevents unexpected behavior if the architecture changes later.

#### **The Fused Loss & Backpropagation Loop**

* **`F.cross_entropy`:** Takes raw logits and applies the Softmax function internally for numerical efficiency and calculation stability.
* **`loss.backward()`**: Computes the gradients along the recorded computational graph.
* **`optimizer.step()`**: Executes the optimization algorithm (e.g., SGD), updating the weights by subtracting the scaled gradients.
* **`optimizer.zero_grad()`**: Clears the gradient history. Skipping this causes PyTorch to accumulate gradients across steps, leading to incorrect calculations and optimization failure.

After we trained the model, we can use it to make predictions, as shown below:

```python
model.eval()

with torch.no_grad():
    outputs = model(X_train)

print(outputs)

torch.set_printoptions(sci_mode=False)
probas = torch.softmax(outputs, dim=1)
print(probas)

predictions = torch.argmax(probas, dim=1)
print(predictions)
```

```python
predictions = torch.argmax(outputs, dim=1)
print(predictions)
```

To generalize the computation of the prediction accuracy, let’s implement a compute_accuracy function as shown in the following code.

```python
def compute_accuracy(model, dataloader):

    model.eval()
    correct = 0.0
    total_examples = 0
    
    for idx, (features, labels) in enumerate(dataloader):
        
        with torch.no_grad():
            logits = model(features)
        
        predictions = torch.argmax(logits, dim=1)
        compare = labels == predictions
        correct += torch.sum(compare)
        total_examples += len(compare)

    return (correct / total_examples).item()
```

```python
compute_accuracy(model, train_loader)
```

The `compute_accuracy` function evaluates a model's performance by processing data through a `DataLoader` rather than loading the entire dataset at once. This approach has a few advantages as follows:

* **Memory Efficiency:** Large datasets cannot fit into GPU/CPU memory simultaneously. Iterating via a data loader ensures the model only processes a small, manageable chunk (batch) at any given moment.
* **Scalability:** This approach allows the evaluation metric to scale seamlessly to datasets of arbitrary size, using the same batch structure defined during training.
* **Consistency:** The internal logic mirrors standard inference, converting raw model logits into predicted class labels batch-by-batch to calculate the final accuracy score.

```python
compute_accuracy(model, test_loader)
```

## **8. Saving and loading models**


PyTorch uses a model's `state_dict` to save and restore its learned parameters (weights and biases) efficiently. The recommended approach is to save only the parameters, not the entire model object. This creates a lightweight Python dictionary containing the model's weights. Standard practice for the file xtension is `.pth` or `.pt`.

```python
torch.save(model.state_dict(), "model.pt")
```

To restore a saved model, we must recreate the empty architecture first and then inject the saved weights into it.

The code structure (`NeuralNetwork(2, 2)`) used during loading must match the original architecture exactly. If we add a layer or change a layer's size, PyTorch will throw a structural mismatch error because the saved weights won't line up with the new layers.
